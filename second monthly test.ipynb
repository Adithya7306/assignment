```python
df['Region']=df['Region'].str.strip().str.lower()
```

PART B

```python
total_sales=df['Sales'].sum()
print('total sales:',total_sales)
```

```python
average_sales=df['Sales'].mean()
print('average sales:',average_sales)
```

```python
category_sales=df.groupby('Category')['Sales'].sum()

highest_category=category_sales.idxmax()
print('Highest-performing category:',highest_category)
```

```python
lowest_category=category_sales.idxmax()
print('Lowest-performing category:',lowest_category)
```

```python
Region_sales=df.groupby('Region')['Sales'].sum()
print(Region_sales)
```

```python
df.select_dtypes(include='number').describe()
```

```python
df.groupby('Category')['Sales'].sum().plot(kind='bar')

plt.title('sales by category')
plt.xlabel('Category')
plt.ylabel('total sales')
plt.show()
```

```python
df['Sales'].plot(kind='hist',bins=20)
plt.title('sales distribution')
plt.xlabel('Category')
plt.ylabel('frequency')
plt.show()
```

```python
df.boxplot(column='Sales')

plt.title('sales boxplot')
plt.ylabel('Sales')
plt.show()
```

```python
plt.scatter(df['Quantity'],df['Sales'])

plt.title('Quantity vs sales')
plt.xlabel('Quantity')
plt.ylabel('Sales')
plt.show()
```

PART C

```python
category_sales=df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
print(category_sales)

print('Highest sales category:',category_sales.idxmax())
```
